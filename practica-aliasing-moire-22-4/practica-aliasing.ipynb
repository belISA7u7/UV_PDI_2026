Reduccion de una imagen con error de muestreo "Aliasing"

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('imagem.png', 0)

plt.figure(figsize=(6,6))
plt.imshow(img, cmap='gray')
plt.title('Imagen con aliasing')
plt.axis('off')
plt.show()
```

```python
gauss = cv2.GaussianBlur(img, (5,5), 0)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original con aliasing')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(gauss, cmap='gray')
plt.title('Suavizado gaussiano')
plt.axis('off')

plt.show()
```

```python
suave2 = cv2.GaussianBlur(gauss, (9,9), 0)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(gauss, cmap='gray')
plt.title('Primer suavizado')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(suave2, cmap='gray')
plt.title('Reducción de altas frecuencias')
plt.axis('off')

plt.show()
```

```python
h, w = suave2.shape

# Aumentar tam con una mejor interpolación
restaurada = cv2.resize(suave2, (w*2, h*2), interpolation=cv2.INTER_CUBIC)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(suave2, cmap='gray')
plt.title('Antes del resize')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(restaurada, cmap='gray')
plt.title('Restaurada con resize correcto')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Cargar imagen con aliasing
img = cv2.imread('imagem.png', 0)

# 2. Suavizado gaussiano
gauss = cv2.GaussianBlur(img, (5,5), 0)

# 3. Reducción adicional de altas frecuencias
suave2 = cv2.GaussianBlur(gauss, (9,9), 0)

# 4. Resize correcto
h, w = suave2.shape
restaurada = cv2.resize(suave2, (w*2, h*2), interpolation=cv2.INTER_CUBIC)

# 5. Mostrar resultados
plt.figure(figsize=(16,4))

plt.subplot(1,4,1)
plt.imshow(img, cmap='gray')
plt.title('Imagen con aliasing')
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(gauss, cmap='gray')
plt.title('Suavizado gaussiano')
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(suave2, cmap='gray')
plt.title('Reducción de altas frecuencias')
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(restaurada, cmap='gray')
plt.title('Restaurada')
plt.axis('off')

plt.show()
```

El aliasing es un efecto visual que produce bordes dentados o escalonados por un muestreo inadecuado de la imagen.
Para disminuir este problema, primero se aplicó un filtro gaussiano que suaviza los cambios bruscos de intensidad.
Después se realizó una reducción adicional de altas frecuencias para atenuar detalles no deseados asociados al aliasing.
Finalmente, se aplicó un redimensionamiento con interpolación cúbica, que mejora visualmente la reconstrucción de los bordes.
El resultado obtenido no recupera información perdida completamente, pero sí reduce de manera notable el efecto de aliasing.
