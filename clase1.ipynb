```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [12, 5]

# Leer imagen
img = cv2.imread('colores.jpeg')
if img is None:
    raise ValueError("No se pudo cargar la imagen")

imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 🎨 función para efecto pastel
def pastel_effect(image, color, strength=0.5):
    pastel = image.copy().astype(np.float32)
    overlay = np.full_like(pastel, color, dtype=np.float32)
    pastel = pastel * (1 - strength) + overlay * strength
    pastel = np.clip(pastel, 0, 255).astype(np.uint8)
    return pastel

# 🌸 10 colores pastel (RGB)
pastel_colors = [
    (255, 182, 193),  # rosa pastel
    (176, 224, 230),  # celeste pastel
    (152, 251, 152),  # verde pastel
    (255, 218, 185),  # durazno
    (221, 160, 221),  # lila
    (255, 239, 213),  # crema
    (240, 230, 140),  # amarillo suave
    (173, 216, 230),  # azul claro
    (216, 191, 216),  # lavanda
    (255, 228, 225),  # rosado claro
]

# Crear figura 2x5
plt.figure(figsize=(15,6))

for i, color in enumerate(pastel_colors):
    img_pastel = pastel_effect(imgRGB, color, strength=0.45)
    
    plt.subplot(2, 5, i+1)
    plt.imshow(img_pastel)
    plt.title(f"Pastel {i+1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np  

plt.rcParams['figure.figsize'] = [12, 6]

img = cv2.imread('colores.jpeg', cv2.IMREAD_COLOR)

# verificar que se cargó
if img is None:
    print("Error: no se pudo cargar la imagen")

imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = imgRGB[:, :, 0]
g = imgRGB[:, :, 1]
b = imgRGB[:, :, 2]

# ---- crear 10 variaciones usando la MISMA lógica ----

imagenes = []

# 1 rojo
img1 = np.zeros_like(imgRGB)
img1[:, :, 0] = r
imagenes.append(("Rojo", img1))

# 2 verde
img2 = np.zeros_like(imgRGB)
img2[:, :, 1] = g
imagenes.append(("Verde", img2))

# 3 azul
img3 = np.zeros_like(imgRGB)
img3[:, :, 2] = b
imagenes.append(("Azul", img3))

# 4 rojo + verde
img4 = np.zeros_like(imgRGB)
img4[:, :, 0] = r
img4[:, :, 1] = g
imagenes.append(("Rojo+Verde", img4))

# 5 rojo + azul
img5 = np.zeros_like(imgRGB)
img5[:, :, 0] = r
img5[:, :, 2] = b
imagenes.append(("Rojo+Azul", img5))

# 6 verde + azul
img6 = np.zeros_like(imgRGB)
img6[:, :, 1] = g
img6[:, :, 2] = b
imagenes.append(("Verde+Azul", img6))

# 7 mitad rojo
img7 = np.zeros_like(imgRGB)
img7[:, :, 0] = r // 2
imagenes.append(("Rojo Suave", img7))

# 8 mitad verde
img8 = np.zeros_like(imgRGB)
img8[:, :, 1] = g // 2
imagenes.append(("Verde Suave", img8))

# 9 mitad azul
img9 = np.zeros_like(imgRGB)
img9[:, :, 2] = b // 2
imagenes.append(("Azul Suave", img9))

# 10 imagen original
imagenes.append(("Original", imgRGB))

# ---- mostrar en 2x5 ----

plt.figure(figsize=(15,6))

for i, (titulo, imagen) in enumerate(imagenes):
    plt.subplot(2, 5, i+1)
    plt.title(titulo)
    plt.imshow(imagen)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np  

plt.rcParams['figure.figsize'] = [12, 6]

img = cv2.imread('colores.jpeg', cv2.IMREAD_COLOR)

if img is None:
    print("Error: no se pudo cargar la imagen")

imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = imgRGB[:, :, 0]
g = imgRGB[:, :, 1]
b = imgRGB[:, :, 2]

imagenes = []

# 1 rojo puro
img1 = np.zeros_like(imgRGB)
img1[:, :, 0] = r
imagenes.append(("Rojo", img1))

# 2 verde puro
img2 = np.zeros_like(imgRGB)
img2[:, :, 1] = g
imagenes.append(("Verde", img2))

# 3 azul puro
img3 = np.zeros_like(imgRGB)
img3[:, :, 2] = b
imagenes.append(("Azul", img3))

# 4 amarillo (rojo + verde)
img4 = np.zeros_like(imgRGB)
img4[:, :, 0] = r
img4[:, :, 1] = g
imagenes.append(("Amarillo", img4))

# 5 magenta (rojo + azul)
img5 = np.zeros_like(imgRGB)
img5[:, :, 0] = r
img5[:, :, 2] = b
imagenes.append(("Magenta", img5))

# 6 cian (verde + azul)
img6 = np.zeros_like(imgRGB)
img6[:, :, 1] = g
img6[:, :, 2] = b
imagenes.append(("Cian", img6))

# 7 naranja (rojo fuerte + verde suave)
img7 = np.zeros_like(imgRGB)
img7[:, :, 0] = r
img7[:, :, 1] = g // 2
imagenes.append(("Naranja", img7))

# 8 violeta (azul fuerte + rojo suave)
img8 = np.zeros_like(imgRGB)
img8[:, :, 0] = r // 2
img8[:, :, 2] = b
imagenes.append(("Violeta", img8))

# 9 verde lima (verde fuerte + rojo suave)
img9 = np.zeros_like(imgRGB)
img9[:, :, 0] = r // 3
img9[:, :, 1] = g
imagenes.append(("Verde Lima", img9))

# 10 turquesa (azul + verde suave)
img10 = np.zeros_like(imgRGB)
img10[:, :, 1] = g // 2
img10[:, :, 2] = b
imagenes.append(("Turquesa", img10))

# ---- mostrar en 2x5 ----

plt.figure(figsize=(15,6))

for i, (titulo, imagen) in enumerate(imagenes):
    plt.subplot(2, 5, i+1)
    plt.title(titulo)
    plt.imshow(imagen)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
