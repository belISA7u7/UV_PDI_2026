Generar imagen RGB con 4 cuadrantes 

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [6, 6]

size = 1000
mitad = size // 2

R = np.zeros((size, size))
G = np.zeros((size, size))
B = np.zeros((size, size))

#OPTENER LOS CUADRANTES DE LA IMAGEN OBTIMIZAR ESTE CODIGO PARA QUE SEA MAS RAPIDO Y EFICIENTE
R[0:mitad, 0:mitad] = 255
G[0:mitad, mitad:size] = 255
B[mitad:size, 0:mitad] = 255
G[mitad:size, mitad:size] = 255
B[mitad:size, mitad:size] = 255

imagen = np.dstack((R, G, B)).astype(np.uint8)

plt.imshow(imagen)
plt.title("Imagen RGB con 4 cuadrantes")
plt.axis("off")
plt.show()
```

Mismo codigo de arriba mas optimizado 

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [6, 6]

size = 1000
mitad = size // 2

R = np.zeros((size, size))
G = np.zeros((size, size))
B = np.zeros((size, size))
#lista de colores para cada cuadrante
colores = [
    (255, 0, 0),    # r
    (0, 255, 0),    # v
    (0, 0, 255),    # a
    (0, 255, 255)   # c
]

# posiciones de cuadrantes
posiciones = [
    (0, mitad, 0, mitad),           # arriba izquierda
    (0, mitad, mitad, size),        # arriba derecha
    (mitad, size, 0, mitad),        # abajo izquierda
    (mitad, size, mitad, size)      # abajo derecha
]

# ciclo optimizado
for (r0, r1, c0, c1), (rv, gv, bv) in zip(posiciones, colores):
    R[r0:r1, c0:c1] = rv
    G[r0:r1, c0:c1] = gv
    B[r0:r1, c0:c1] = bv

imagen = np.dstack((R, G, B)).astype(np.uint8)

plt.imshow(imagen)
plt.title("Imagen optimizada con for")
plt.axis("off")
plt.show()
```

Cubo rubik con colores en cada cara

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [6, 6]

size = 1500
n = 3
bloque = size // n  


R = np.zeros((size, size))
G = np.zeros((size, size))
B = np.zeros((size, size))

colores = [
    (255, 0, 0),      # rojo
    (0, 255, 0),      # verde
    (255, 255, 255),  # blanco
    (230, 150, 0),    # naranja
    (0, 0, 255),      # azul
    (255, 255, 0),    # amarillo
    (255, 0, 0),      # rojo
    (0, 255, 0),    # verde
    (0, 0, 255)     # azul
]

k = 0

# doble ciclo para llenar cada bloque con el color correspondiente
for i in range(n):
    for j in range(n):

        r0 = i * bloque 
        r1 = (i + 1) * bloque
        c0 = j * bloque
        c1 = (j + 1) * bloque

        rv, gv, bv = colores[k]

        R[r0:r1, c0:c1] = rv
        G[r0:r1, c0:c1] = gv
        B[r0:r1, c0:c1] = bv

        k += 1

imagen = np.dstack((R, G, B)).astype(np.uint8)

plt.imshow(imagen)
plt.title("Matriz 3x3 tipo Cubo Rubik")
plt.axis("off")
plt.show()
```
