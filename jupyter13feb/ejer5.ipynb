```python


numeros = []

for i in range(7):
    num = int(input(f"Ingrese el número {i+1}: "))
    numeros.append(num)

repeticiones = {}

for num in numeros:
    if num in repeticiones:
        repeticiones[num] += 1
    else:
        repeticiones[num] = 1

mayor_repetido = max(repeticiones, key=repeticiones.get)

print(f"El número que más se repite es {mayor_repetido}")
print(f"Se repite {repeticiones[mayor_repetido]} veces")
```
