Solarizacion

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 
def solarizar(imagen, umbral):
    # Aplicamos SOL
    img_sol=np.where(imagen < umbral, imagen, 255 - imagen)
    return img_sol

imagen=cv2.imread('img3.png')
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
umbral=128
#---
img_solarizada=solarizar(img_gray, umbral)

plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title('Imagen Original')    
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.subplot(1,2,2)
plt.title('Imagen Solarizada')
plt.imshow(img_solarizada, cmap='gray')
```

Posterilizar

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

def posterizar(imagen, niveles):
    # Aplicamos POSTERIZAR
    factor_escalado = 256 // niveles
    img_posterizada =np.round(imagen / factor_escalado) * factor_escalado
    return img_posterizada.astype(np.uint8)


imagen=cv2.imread('img3.png')
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
niveles=5
#---
img_posterizada=posterizar(img_gray, niveles)   
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.subplot(1,2,2)
plt.title('Imagen Posterizada')
plt.imshow(img_posterizada, cmap='gray')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def posterizar(imagen,niveles):
    factor_escala= 255/niveles
    img_pos=np.round(imagen/factor_escala)*factor_escala
    return img_pos.astype(np.uint8)

def solarizar(imagen,umbral):
    #APLICAMOS SOL
    img_sol=np.where(imagen<umbral,imagen,255-imagen)
    return img_sol

imagen = cv2.imread("img3.png")
umbral = 127
img_fin = solarizar(imagen,umbral)
nvl=1.5
img_fin2 = posterizar(imagen,nvl)
plt.figure(figsize=(12,6))
plt.subplot(131)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.subplot(132)
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))
plt.subplot(133)
plt.imshow(cv2.cvtColor(img_fin2, cv2.COLOR_BGR2RGB))
plt.show()
```

Falso Color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

def falso_color(img_gray):
    
    # Aplicamos FALSO COLOR
    escala_img= cv2.normalize(img_gray, None, 0, 255, cv2.NORM_MINMAX)
    img_falso_color = cv2.applyColorMap(escala_img, cv2.COLORMAP_MAGMA)
    return img_falso_color

imagen=cv2.imread('img3.png')
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
#---
img_falso_color=falso_color(img_gray)
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.subplot(1,2,2)
plt.title('Imagen con Falso Color')
plt.imshow(img_falso_color)
plt.show()
```
