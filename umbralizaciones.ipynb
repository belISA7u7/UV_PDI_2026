## Umbralizaciones binaria y binaria inversa 

```python
##convertir la imagen en escala de grises
import cv2
import matplotlib.pyplot as plt
import numpy as np
imagen = cv2.imread('colores.jpeg', 0) #lee en escala de grises la imagen
#umbralizacion binaria
_,umbral_binario = cv2.threshold(imagen,100,255, cv2.THRESH_BINARY)
#mostrar la imagen original y la umbralizada inversa
_,umbral_binaria_inv = cv2.threshold(imagen,100,255, cv2.THRESH_BINARY_INV)

plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(imagen, cmap='gray')
plt.title('Imagen Original')

plt.subplot(1,3,2)
plt.imshow(umbral_binario, cmap='gray') 
plt.title('Umbralización Binaria')
plt.subplot(1,3,3)

plt.imshow(umbral_binaria_inv, cmap='gray')
plt.title('Umbralización Binaria Inversa')
plt.tight_layout()
```
